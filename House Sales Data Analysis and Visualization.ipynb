```python
#1
visual_1=df.groupby("bedrooms")["price"].count()
visual_1
```

```python
import matplotlib.pyplot as plt
import numpy as np

# GIVEN ORDER: 0 TO 9 BEDROOMS
visual_1 = visual_1.reindex(range(10))

# 5 DARK PURPLE FAMILY SHADES
purple_colors = [
    "#E6D5F7",   # LIGHTEST
    "#C39BE8",
    "#9B62D0",
    "#7139A8",
    "#3D1766"    # DARKEST
]

# Assign shades based on values
values = visual_1.values
rank = np.argsort(np.argsort(values))

# Convert rank into 5 colour levels
color_index = (rank * 5 // len(values)).clip(0, 4)

colors = [purple_colors[i] for i in color_index]

# Plot
ax = visual_1.plot(
    kind="bar",
    figsize=(11, 6),
    color=colors,
    width=0.65,
    edgecolor="#2B1048",
    linewidth=0.8
)

# TITLE
plt.title(
    "HOUSE DISTRIBUTION BY BEDROOMS",
    fontsize=17,
    fontweight="bold",
    fontfamily="DejaVu Sans",
    pad=15
)

# AXIS LABELS
plt.xlabel(
    "NUMBER OF BEDROOMS",
    fontsize=11,
    fontfamily="DejaVu Sans"
)

plt.ylabel(
    "NUMBER OF HOUSES",
    fontsize=11,
    fontfamily="DejaVu Sans"
)

# KEEP 0 TO 9 ORDER
plt.xticks(
    range(10),
    [str(i) for i in range(10)],
    rotation=0,
    fontsize=10
)

plt.yticks(fontsize=10)

# VALUE LABELS
for container in ax.containers:
    ax.bar_label(
        container,
        fmt="%.0f",
        padding=3,
        fontsize=9
    )

# CLEAN GRID
plt.grid(
    axis="y",
    linestyle="--",
    alpha=0.2
)

plt.tight_layout()
plt.show()
```

```python
month=["May","June","July"]
visual_9=df.groupby("MONTH")["price"].sum().sort_values()
visual_9=visual_9.reindex(month)
visual_9
```

```python
import matplotlib.pyplot as plt
import numpy as np

colors = ["#D1B3E8", "#A875C9", "#7B45A8"]

fig, ax = plt.subplots(
    1, 2,
    figsize=(14, 6)
)

# ================= BAR CHART FIRST =================

visual_9.plot(
    kind="bar",
    ax=ax[0],
    color=colors,
    width=0.35,
    edgecolor="#2B1048",
    linewidth=0.8
)

ax[0].set_title(
    "TOTAL HOUSE PRICE BY MONTH",
    fontsize=16,
    fontweight="bold",
    fontfamily="DejaVu Sans"
)

ax[0].set_xlabel("MONTH")
ax[0].set_ylabel("TOTAL HOUSE PRICE")
ax[0].tick_params(axis="x", rotation=0)

# SHOW VALUES
for container in ax[0].containers:
    ax[0].bar_label(
        container,
        fmt="%.2f",
        padding=3,
        fontsize=9
    )

ax[0].grid(
    axis="y",
    linestyle="--",
    alpha=0.2
)


# ================= PIE CHART SECOND =================

wedges, _ = ax[1].pie(
    visual_9,
    colors=colors[:len(visual_9)],
    startangle=90,
    wedgeprops={
        "edgecolor": "white",
        "linewidth": 1.5
    }
)

for i, wedge in enumerate(wedges):

    angle = (wedge.theta1 + wedge.theta2) / 2

    x = np.cos(np.radians(angle))
    y = np.sin(np.radians(angle))

    percentage = visual_9.iloc[i] / visual_9.sum() * 100

    ax[1].annotate(
        f"{str(visual_9.index[i]).upper()}\n{percentage:.2f}%",
        xy=(x * 0.9, y * 0.9),
        xytext=(x * 1.35, y * 1.35),
        ha="center",
        va="center",
        fontsize=10,
        arrowprops=dict(
            arrowstyle="->",
            connectionstyle="arc3,rad=0.15"
        )
    )

ax[1].set_title(
    "TOTAL HOUSE PRICE BY MONTH",
    fontsize=16,
    fontweight="bold",
    fontfamily="DejaVu Sans"
)

ax[1].set_ylabel("")

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

ax = visual_9.plot(
    kind="line",
    figsize=(10, 5),
    marker="o",
    linewidth=2,
    color="#6A0DAD"
)

plt.title(
    "PRICE PER MONTH",
    fontsize=16,
    fontweight="bold",
    fontfamily="DejaVu Sans",
    pad=15
)

plt.xlabel("MONTH", fontsize=11)
plt.ylabel("PRICE", fontsize=11)

plt.xticks(rotation=0, fontsize=10)
plt.yticks(fontsize=10)

# VALUES SLIGHTLY ABOVE
for x, y in enumerate(visual_9):
    plt.text(
        x, y + 300,
        f"{y:,.0f}",
        ha="center",
        va="bottom",
        fontsize=9
    )

plt.grid(
    axis="y",
    linestyle="--",
    alpha=0.2
)

plt.tight_layout()
plt.show()
```

```python
#10
day=["Sunday","Monday","Tuesday","Wednesday","Thursday","Friday","Saturday"]
visual_10=df.groupby("DAY")["price"].sum().sort_values()
visual_10=visual_10.reindex(day)
visual_10
```

```python
import matplotlib.pyplot as plt
import numpy as np

# HIGHER VALUE = DARKER PURPLE
rank = visual_10.rank(method="min", ascending=True)

colors = plt.cm.Purples(
    np.interp(
        rank,
        [rank.min(), rank.max()],
        [0.35, 0.90]
    )
)

fig, ax = plt.subplots(
    1, 2,
    figsize=(15, 7)
)

# ================= BAR CHART FIRST =================

visual_10.plot(
    kind="barh",
    ax=ax[0],
    color=colors,
    width=0.35,
    edgecolor="#2B1048",
    linewidth=0.8
)

ax[0].set_title(
    "TOTAL HOUSE PRICE BY DAY",
    fontsize=16,
    fontweight="bold",
    fontfamily="DejaVu Sans"
)

ax[0].set_xlabel("TOTAL HOUSE PRICE")
ax[0].set_ylabel("DAY")

ax[0].tick_params(axis="x", rotation=45)

# SHOW VALUES
for container in ax[0].containers:
    ax[0].bar_label(
        container,
        fmt="%.2f",
        padding=3,
        fontsize=9,
        rotation=0
    )

ax[0].grid(
    axis="y",
    linestyle="--",
    alpha=0.2
)


# ================= DONUT CHART SECOND =================

wedges, _ = ax[1].pie(
    visual_10,
    colors=colors,
    startangle=90,
    wedgeprops={
        "width": 0.40,
        "edgecolor": "white",
        "linewidth": 1.5
    }
)

# DAY + ACTUAL VALUE OUTSIDE
for i, wedge in enumerate(wedges):

    angle = (wedge.theta1 + wedge.theta2) / 2

    x = np.cos(np.radians(angle))
    y = np.sin(np.radians(angle))

    value = visual_10.iloc[i]

    ax[1].annotate(
        f"{str(visual_10.index[i]).upper()}\n{value:,.2f}",
        xy=(x * 0.9, y * 0.9),
        xytext=(x * 1.40, y * 1.40),
        ha="center",
        va="center",
        fontsize=9,
        arrowprops=dict(
            arrowstyle="->",
            connectionstyle="arc3,rad=0.15"
        )
    )



# TOTAL IN DONUT CENTER
total_price = visual_10.sum()

ax[1].text(
    0,
    0,
    f"TOTAL\n{total_price:,.2f}",
    ha="center",
    va="center",
    fontsize=11,
    fontweight="bold"
)

ax[1].axis("equal")

plt.tight_layout()
plt.show()
```
